## 1. Lloyd Shapley: Kooperatif Oyun Kuramı ve 4 Aksiyom İspatı

Lloyd Shapley'nin 1953 tarihli teoremine göre, kooperatif bir oyunda adil ödül dağılımını garanti eden tek formülasyon Shapley değeridir:

$$\phi_i(v) = \sum_{S \subseteq N \setminus \{i\}} \frac{|S|!(|N|-|S|-1)!}{|N|!} [v(S \cup \{i\}) - v(S)]$$

Aşağıda analitik bir oyunda 4 aksiyomun (Verimlilik, Simetri, Etkisiz Eleman, Toplamsallık) sağlandığı doğrulanmaktadır.

```python
game = create_axiomatic_game(n_players=4)
report = verify_all_shapley_axioms(game)

print(report['summary'])
print('\nOyuncuların Kesin Shapley Değerleri:')
for name, val in zip(game.player_names, report['shapley_values']):
    print(f'  - {name:<15}: {val:.4f}')
```

## 2. John Maynard Smith: Evrimsel Oyun Teorisi ve Replicator Dinamikleri

Replicator dynamics denklemi:
$$\dot{x}_i = x_i [(A x)_i - x^T A x]$$

Popülasyonda güçlü ve sağlam stratejilerin frekansı zamanla artarken, kırılgan stratejiler yok olur.

```python
payoff_matrix = np.array([
    [-1.0, 4.0],
    [0.0, 2.0]
])

traj, times = simulate_replicator_dynamics(payoff_matrix, initial_shares=[0.9, 0.1], generations=120)

plt.figure(figsize=(9, 4))
plt.plot(times, traj[:, 0], label='Strateji 1 (Agresif / Şahin)', color='#e74c3c', lw=2)
plt.plot(times, traj[:, 1], label='Strateji 2 (Savunmacı / Güvercin)', color='#3498db', lw=2)
plt.axhline(2/3, color='black', linestyle='--', label='Teorik ESS Denge Noktası (2/3)')
plt.title('Maynard Smith Replicator Dynamics Evrimsel Denge Süreci', fontweight='bold')
plt.xlabel('Zaman (t)')
plt.ylabel('Popülasyon Oranı')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
